```python
from datetime import datetime, timedelta
import time

hora = datetime.now()

print(f"Tipo de objeto de la hora {type(hora)}")
print(f"La hora es {hora}")
print(f"La hora de {hora} es {hora.hour}")
print(f"El minuto de {hora} es {hora.minute}")
print(f"El segundo de {hora} es {hora.second}")
print(f"El microsegundo de {hora} es {hora.microsecond}")
```

```python
fActual = datetime.today()
print(f"Tipo de objeto de la fecha actual {type(fActual)}")
print(f"La fecha actual es {fActual}")
print(f"El año actual es {fActual.year}")
print(f"El mes actual es {fActual.month}")
print(f"El día actual es {fActual.day}")
```

```python
fecha_cap = input("Fecha (dd/mm/aaaa): ")
fecha_obj = datetime.strptime(fecha_cap, "%d/%m/%Y")
print(f"Tipo de objeto de la fecha capturada {type(fecha_cap)}")
print(f"Tipo de objeto de la fecha procesada {type(fecha_obj)}")
print(f"La fecha capturada pasó a {fecha_obj}")
```

```python
print(f"Tenemos {fecha_obj} en la variable con el tipo de dato especializado")
print(f"El mismo valor lo podemos presentar en el formato dd/mm/aaaa con {fecha_obj.strftime('%d/%m/%Y')}")
print(f"O tambien en el formato mm/dd/aaaa: {fecha_obj.strftime('%m/%d/%Y')}")
```

```python
cantDias = int(input(f"Dime la cant de dias a adelantar con respecto a {fecha_obj}"))
nFecha = fecha_obj + timedelta(days=+cantDias)
print(f"La fecha adelantada es {nFecha}")
```
