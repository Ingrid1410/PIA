```python
import sys

lista_valores = ['x', 0, 4]

for valor in lista_valores:
  try:
    print(f"El valor en turno es {valor}")
    reciproco = 1 / int(valor)
    print(f"El reciproco es {reciproco}")
  except Exception:
    print(f"Ocurrio un problema {sys.exc_info()[0]}")
    print("Procedemos al siguiente valor")
```

```python
import sys
lista_valores = ['x', 0, 4]

for valor in lista_valores:
  try:
    print(f"El valor en turno es {valor}")
    reciproco = 1 / int(valor)
    print(f"El reciproco es {reciproco}")
  except ValueError:
    print(f"El valor proporcionado {valor} no es compatible con la operacion solicitada")
  except ZeroDivisionError:
    print("La division sobre zero no esta permitida")
  except Exception:
    print(f"Ocurrio un problema {sys.exc_info() ^ }")
```
