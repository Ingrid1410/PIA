```python
import random
#Demostracion de randrange(), randint() y random()

print(f"Obteniendo un numero aleatorio que pueda ir del 0 al 19: {random.randrange(20)}")

print(f"Obteniendo un numero entero aleatorio par que pueda ir de 2 al 20: {random.randrange(2, 21, 2)}")

print(f"Obteniendo un numero entero aleatorio que pueda ir del 1 hasta el 20: {random.randint(1, 20)}")

print(f"Obteniendo un valor numerico entre 0.0 y 1.0 inclusive: {random.random()}")
```

```python
#Demostracion de choice() y shuffle()
listaDePrueba = [10, 20, 30, 40, 15, 25, 35, 45]
print(f"La lista de prueba es {listaDePrueba}")

print(f"El valor seleccionado aleatoriamente de la lista anterior es {random.choice(listaDePrueba)}")

random.shuffle(listaDePrueba)
print(f"La lista ya 'perturbada/bajada' es {listaDePrueba}")
```
