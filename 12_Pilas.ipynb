```python
#Implementacion de una pila en python

class Stack:    #Creamos la clase stack
  def __init__(self):
    self.items = []
  def is_empty(self):
    return self.items == []
  def push(self, item):
    self.items.insert(0, item)
  def pop(self):
    return self.items.pop(0)
  def print_stack(self):
      print(self.items)
pila = Stack()

pila.push('a')
pila.push('b')
pila.push('c')
pila.push('d')
pila.push('e')
pila.print_stack()
pila.pop()
pila.print_stack()
pila.pop()
pila.print_stack()
```

```python
class Pila:
  def __init__(self):
    self.items = []

  def estaVacia(self):
    return self.items == []

  def extraer(self):
    return self.items.pop()

  def incluir(self, item):
    self.items.append(item)

s = Pila()
s.incluir('hola')
s.incluir('verdadero')
s.incluir(1)
s.incluir('falso')
print(s.extraer())
```

```python
stack = []
SEPARA = ('-' * 20) + "\n"

def main():
  print("1 Apillar elemento (entero)")
  print("2 Desapilar elemento")
  print("3 Mostrar pila") # Added option 3, as it's missing but implied by the other options
  print("4 Salir")
  print(SEPARA)

option = input("Elija una opcion: ")

if str(option)=="1":
    elemento = input(" Introduzca el numero a apilar: ")
    stack.append(elemento)
    print("''- Elemento apilado -''")
    main()

elif str(option)== "2":
    if len(stack) == 0:
        print(" No hay elementos para desapilar ")
        main()
    else:
        print("''-El numero: " + str(stack.pop()) + " Fue desapilado-''")
        main()

elif str(option)== "3": # Handling the 'Mostrar pila' option
    if len(stack) == 0:
        print(" La pila esta vacia ")
    else:
        print("Pila actual: ", stack)
    main()

elif str(option)=="4":
    exit()

else:
    print("Opcion incorrecta.\n")
    main()
```

```python
#Primera opcion: Utilizar una lista

pila_con_lista = list()
for _ in range(5):
  pila_con_lista.append(input("Dime el nombre a agregar: "))
print(f"Se han agregado {len(pila_con_lista)} elementos a la pila implementada con una lista\n")

#Sacar elementos de la pila
while pila_con_lista:
  print(pila_con_lista.pop())
print(f"Actualmente quedan {len(pila_con_lista)} elementos de la pila implementada con una lista\n")
```

```python
#Segunda opcion: Utilizar un objeto deque

import collections

pila_con_deque = collections.deque()
for _ in range(5):
  pila_con_deque.append(input("Dime el nombre a agregar: "))
print(f"Se han agregado {len(pila_con_deque)} elementos a la pila implementada con una lista\n")

while pila_con_deque:
  print(pila_con_deque.pop())
print(f"Actualmente quedan {len(pila_con_deque)} elementos a la pila implementada con una lista\n")
```

```python
class cola:
  def __init__(self):
    self.items = []
  def estaVacia(self):
    return self.items == []
  def agregar(self, item):
    self.items.insert(0,item)
  def avanzar(self):
    return self.items.pop()
  def tamano(self):
    return len(self.items)

c=cola()

c.agregar(4)
c.agregar('perro')
c.agregar(True)
c.agregar()
```

```python
#Creamos una lista vacia
cola = []
SEPARA = ("-" * 20) + "\n" #Se crea variable para separar cada menu, \n es salto de linea

#Creamos un menu con 4 opciones
def main():
  print("1 Insertar cola")
  print("2 Borrar en cola")
  print("3 Listar cola")
  print("4 Salir")
  print(SEPARA * 1)

  option = input("Elija una opcion")
  if str(option)=="1":
    elemento = input("Introduzca el numero a encolar: ")
    cola.append(elemento)
    print(" numero encolado con exito")
    main()

  elif str(option)=="2":
    if len(cola)>0:
      print("El numero:",cola.pop(0), "fue descolado")
      main()
    else:
        print("Cola vacia")
        main()

  elif str(option)=="3":
    for i in cola:
      print("cola: ", i)
    main()
  elif str(option)=="4":
    exit()

  else:
    print("Seleccione una opcion valida.")
    main()

main()
```
