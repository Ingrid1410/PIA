```python
import time

segundos = int(input("Cantidad der segundos a esperar:\n"))
time.sleep(segundos) #Espere por lo menos la cantidad de segundos especificada
print(f"Han transcurrido, por lo menos {segundos} segundos")
```

```python
"""Iniciaremos la medicion de un proceso simulado, que espera 10 veces el tiempo que indicamos en la variable segundos de la celda anterior"""

horaInicial = time.time()

for termino in range(10):
  time.sleep(segundos)

print("proceso simulado concluido")
duracion = time.time() - horaInicial  #Puede verse afectado si se cambia la hora del sistema
print(f"La duracion del proceso simulado fue {duracion: .4f} segundos")
```
