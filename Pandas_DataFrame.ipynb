```python
import pandas as pd
SEPARADOR = ("*" * 60) + "\n"

diccionario_notas = {"Programación":[87, 80, 80, 20, 100],
                     "Base de datos": [71, 100, 70, 100, 100],
                     "Contabilidad": [99, 57, 57, 100, 100]}
notas_diccionario = pd.DataFrame(diccionario_notas, index=["Crescencio", "Domitila", "Rutilio", "Panfilo", "Ludoviko"])
print(notas_diccionario)
print(SEPARADOR)
```

```python
print("\nLas notas obtenidas en programación fueron:")
print(notas_diccionario["Programación"])
print()

print(notas_diccionario.Programación)
print(SEPARADOR)
```

```python
print("Las notas obtenidas por Rutilio fueron: ")
print(notas_diccionario.loc["Rutilio"])
print("\nLas notas obtenidas por Ludiviko fueron:")
print(notas_diccionario.loc["Ludoviko"])
```

```python
print(f"\nLa menor calificación de Crescencio fue de {notas_diccionario.loc['Crescencio'].min()}")
print()
print(f"El promedio de crescencio fue de {notas_diccionario.loc['Crescencio'].mean():.4f}")
```

```python
medias = notas_diccionario.mean()
print(type(medias))
print("\nPromedio por UA")
print("Se obtuvo la media de las siguientes asignaturas:")

for materia in medias.index:
  print(materia)

print(f"\nEl promedio de las notas obtenidas fue de")
print(medias)
```

```python
media_global = notas_diccionario.mean().mean()
print(f"\nLa media global del DataFrame es de {media_global:.4f}")
```

```python
print("\nLas notas obtenidas en programación fueron:")
print(notas_diccionario["Programación"])
print()

print(notas_diccionario.Programación)
print(SEPARADOR)
```

```python
print("\nÚltimas dos asignaturas para todos los estudiantes a partir de Rutilio")
print(notas_diccionario[["Base de datos", "Contabilidad"]].loc["Rutilio":])

print(type(notas_diccionario[["Base de datos", "Contabilidad"]].loc["Rutilio":]))
print(SEPARADOR)
```

```python
print("La calificación de Pánfilo en Programación es de")
print(notas_diccionario.at["Panfilo", "Programación"])
```

```python
filtro = notas_diccionario > 79

print("Este sería el filtro o mapa de bits resultante de aplicar la condición")
print(filtro)
```

```python
print("Este sería el filtro o mapa de bits resultante de aplicar la condición")
print(notas_diccionario[filtro])
```

```python
print(notas_diccionario[notas_diccionario < 70])
print(SEPARADOR)

print(notas_diccionario[notas_diccionario < 70].min())
```

```python
base_datos_70_71 = notas_diccionario.loc[notas_diccionario["Base de datos"].isin([70,71])]
print(base_datos_70_71)
```

```python
filtro_aprobados_menor_100 = (notas_diccionario["Programacion"] >= 70) & (notas_diccionario["Programacion"]) < 100
aprobaron_progra_menos_que_100 = notas_diccionario.loc[filtro_aprobados_menor_100]

print(aprobaron_progra_menos_que_100)
print(SEPARADOR)
print(aprobaron_progra_menos_que_100["Programacion"])
```
