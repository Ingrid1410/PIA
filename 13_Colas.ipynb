```python
cola = list()

for cantidad in range(5):
  nuevo = input("Nombre del recien llegado: ")
  cola.append(nuevo)

print(f"Se agregaron {len(cola)} elementos: ")
for elemento in cola:
  print(elemento)

#Hacer notar que los elementos permanecen en la cola
print(f"Actualmente hay {len(cola)} encolados")

print("Procedemos a retirarlos de la cola")

while cola:
  print(cola.pop(0))

#Verificar que la estructura se encuentra vacia
print(f"Actualmente hay {len(cola)} encolados")
```

```python
import collections
cola = collections.deque() #Cola utilizando deque

for cantidad in range(5):
  nuevo = input("Nombre del recien llegado: ")
  cola.append(nuevo)

print(f"Se agregaron {len(cola)} elementos: ")
for elemento in cola:
  print(elemento)

print("Procesamos a retirarlos de la cola")

while cola:
  print(cola.popleft())

print(f"Actualmente se encuentran en la cola {len(cola)} elementos")
```
